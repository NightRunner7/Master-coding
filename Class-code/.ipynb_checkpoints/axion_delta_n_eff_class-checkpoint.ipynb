## Axuiliary functions

```python
def calculate_fa(ma):
    """
    Calculate the f_a value from a given axion mass m_a in eV based on the axion mass
    formula from particle physics. The relationship between the axion decay constant f_a
    and the axion mass m_a is given by the specific scale factor derived from theoretical
    models which predict f_a inversely proportional to m_a.

    This specific formula is referenced from a particle physics review publication:
    https://pdg.lbl.gov/2023/reviews/rpp2023-rev-axions.pdf

    Parameters:
    ma (float): The mass of the axion in eV.

    Returns:
    float: The axion decay constant f_a in GeV.
    """
    fa = (5.691 * 10 ** 6) / ma  # [GeV]
    return fa  # [GeV]
```

## Define parameters of distribution (for axions): $A(\log(m_a)), \, b(\log(m_a)), \,\mu(\log(m_a))$

```python
# Polynomial for A parameter
def A(x):
    # x = np.log(m_a)
    # Coefficients for A(x) (from highest degree to constant term)
    coeff_A = [-7.369e-06, -6.597e-05, -0.0001348, 0.0004274, 0.001855, 
               5.068e-05, -0.007855, -0.006923, 0.01576, 0.02598, -0.01983, -0.08118, 0.9416]
    return np.polyval(coeff_A, x)

# Polynomial for b parameter
def b(x):
    # x = np.log(m_a)
    # Coefficients for b(x) (from highest degree to constant term)
    coeff_b = [-0.0001591, -0.0005741, 0.002432, 0.01034, -0.01059, 
               -0.06682, 0.007572, 0.2287, 0.07249, -0.7259, 0.6496, 0.5267]
    return np.polyval(coeff_b, x)

# Polynomial for mu parameter
def mu(x):
    # x = np.log(m_a)
    # Coefficients for mu(x) (from highest degree to constant term)
    coeff_mu = [-4.667e-05, 0.0006367, 0.002832, -0.003391, -0.01729, 
                -0.006905, 0.04416, 0.01063, 0.1161, -0.2879, 0.1581, -0.1738, 0.576]
    return np.polyval(coeff_mu, x)
```

```python
# --- Deal with A
def A(x):
    """
    Piecewise function combining a plateau and a rational function to model parameter A.
    
    Parameters:
    - x: The input value (or array of values).
    
    Returns:
    - The value of A(x) based on the input.
    """
    # Best fit parameters
    best_parameters_A = [
        1.04720071,  # c: Plateau value for x < x_switch
        0.01126367, 0.03679986, 0.32768154, 1.4278752, 2.02783514, 2.12249276, 3.06914506,  # Numerator coefficients
        0.01312141, 0.0611571, 0.43659234, 1.70826943, 2.56251827, 2.72833247, 3.290127,  # Denominator coefficients
        -2.0  # x_switch: Threshold for switching from plateau to rational function
    ]
    
    # Unpack parameters for readability
    c, a6, a5, a4, a3, a2, a1, a0, b6, b5, b4, b3, b2, b1, b0, x_switch = best_parameters_A
    
    # Define the rational function part
    numerator = a6 * x**6 + a5 * x**5 + a4 * x**4 + a3 * x**3 + a2 * x**2 + a1 * x + a0
    denominator = b6 * x**6 + b5 * x**5 + b4 * x**4 + b3 * x**3 + b2 * x**2 + b1 * x + b0
    rational_part = numerator / (denominator + 1e-10)  # Adding a small value to avoid division by zero
    
    # Piecewise function combining plateau and rational function
    return np.where(x < x_switch, c, rational_part)


# --- Deal with b
def b(x):
    """
    Computes the value of b(x) based on the input x according to the following rules:
    - If x > x_max: b(x) = b_max
    - If x < x_min: b(x) = a * x + b_intercept
    - Otherwise: b(x) = polynomial of degree 11
    
    Parameters:
    - x (float or array-like): Input value(s) for which to calculate b(x), x = log(m_a)
    
    Returns:
    - float or np.ndarray: Calculated value of b(x).
    """
    # Best fit parameters for b
    best_parameters_b = [
        -2.77576175,  # x_min: Threshold value below which b(x) is linear
        1.82940843,   # x_max: Threshold value above which b(x) is constant
        2.02864447,   # a: Slope of the linear function when x < x_min
        1.63986335,   # b_intercept: Intercept of the linear function when x < x_min
        0.52656812,   # b_max: Constant value of b(x) when x > x_max
        [-1.44200678e-04, -6.82799533e-04, 1.38359514e-03, 1.01159544e-02, -1.33010324e-03, -5.99591118e-02, 
         -2.87608043e-02, 2.02285841e-01, 1.60223288e-01, -6.83486489e-01, 4.47212203e-01, 4.89066026e-01]
        # Coefficients of the polynomial of degree 11 (highest degree first)
    ]
    
    # Unpack parameters for readability
    x_min, x_max, a, b_intercept, b_max, polynomial_coeffs = best_parameters_b
    
    # Convert the input x to a numpy array for vectorized operations
    x = np.asarray(x)
    
    # Define the polynomial function of degree 11
    polynomial = np.poly1d(polynomial_coeffs)
    
    # Define b(x) based on the specified conditions
    b_x = np.where(
        x > x_max,                       # Condition: x > x_max
        b_max,                           # Result if true: b(x) = b_max
        np.where(                        # Condition: x < x_max
            x < x_min,                   # Condition inside np.where: x < x_min
            a * x + b_intercept,         # Result if true: b(x) = a * x + b_intercept
            polynomial(x)                # Result if false: b(x) = polynomial of degree 11
        )
    )
    
    return b_x
    
# --- Deal with mu
def mu(x):
    """
    Computes the value of mu(x) using an exponential decay function:
    
    mu(x) = a * exp(-x / b) + c
    
    Parameters:
    - x (float or array-like): Input value(s) for which to calculate mu(x), x = log(m_a), x = log(m_a)
    
    Returns:
    - float or np.ndarray: Calculated value of mu(x).
    """
    # Best fit parameters for mu
    best_parameters_mu = [
        0.29224341,  # a: Controls the amplitude of the decay
        0.50440147,  # b: Controls the rate of decay
        0.13463436   # c: Adjusts the plateau level
    ]
    
    # Unpack parameters for readability
    a, b, c = best_parameters_mu
    
    # Calculate mu(x) using the exponential decay formula
    return a * np.exp(-x / b) + c
```

## Find $T_{\text{ncdm}}(m_a)$: for now I m setting $T_{\text{ncdm}}(m_a) = 0.71611$

```python
def delta_n_eff_and_axion_mass(x):
    """
    Evaluate the 7th degree polynomial for Delta N_eff as a function of x.

    The polynomial is:
    0.1625 * x^7 - 1.218 * x^6 + 3.448 * x^5 - 4.139 * x^4 + 
    0.8498 * x^3 + 1.949 * x^2 + 0.01227 * x - 0.0002925
    """
    # Coefficients of the polynomial
    coeffs = [0.1625, -1.218, 3.448, -4.139, 0.8498, 1.949, 0.01227, -0.0002925]
    
    # Evaluate the polynomial using NumPy's polyval function
    return np.polyval(coeffs, x)
```

```python
print("Delta N_eff(1.0):", delta_n_eff_and_axion_mass(1.0))
```

```python
def calculate_T_axions(axion_mass, netrino_temperature):
    # find \Delta N_eff
    delta_n_eff = delta_n_eff_and_axion_mass(axion_mass)
    # calculate temperature for axions
    T_a = (7/4 * delta_n_eff) ** (1/4) * netrino_temperature
    return T_a
```

## Base parameters setting (we do not vary them)

```python
# Define the cosmological parameters for LambdaCDM + neutrinos + axions
params = {
    'output': 'mPk',  # Output: matter power spectrum
    'P_k_max_1/Mpc': 3.0,  # Maximum wavenumber for the power spectrum
    'h': 0.67556,          # Reduced Hubble parameter
    'omega_b': 0.022032,   # Baryon density
    'omega_cdm': 0.12038,  # Cold dark matter density
    'n_s': 0.9619,         # Spectral index
    'A_s': 2.215e-9,       # Amplitude of scalar perturbations
    'tau_reio': 0.0925,    # Optical depth to reionization

    # Neutrino and axion related parameters
    'N_ur': 0.0,
    'N_ncdm': '2',              # two species: neutrinos and axions
    'deg_ncdm': '3, 1',         # 3 degenerate neutrinos and axion

    # Basic settings
    # 'use_ncdm_psd_files': '0, 0',
    
    # Extra settings
    'perturbed_recombination': 'yes',       # consider perturbed recombination
    'ceff2_ur': 1/3,                        # effective squared sound speed and viscosity parameter
    'cvis2_ur': 1/3,                        # effective squared sound speed and viscosity parameter
    'ncdm_quadrature_strategy': '0, 0',      # is the method used for the momentum sampling of the ncdm distribution function: 0, 1, 2, 3
    # 'ncdm_maximum_q': '15, 15'
}
```

```python
# the range of parameters, which we gonna investigate
Nums = 100
axion_masses = np.logspace(np.log10(0.04), np.log10(2), num=Nums)  # [eV]
nuetrino_mass = 0.006  # [eV]
T_neutrinos = 0.71611  # in units of photon temperature [T_gamma]
# print(axion_masses)

# prepare list
Neff_base = 3.044
deltaNeff_list = []
```

## Varied parameters (axion masses and distribution)

```python
cosmo_dict = {}
#
for i in range(0, Nums):
    m_a = axion_masses[i]  # [eV]
    # T_axion = calculate_T_axions(m_a, T_neutrinos)  # in units of photon temperature [T_gamma]
    # T_axion = 0.68481  # in units of photon temperature [T_gamma]
    T_axion = 0.71611 * (g_s_star_ratio)**(1/3) # in units of photon temperature [T_gamma]
    # T_axion = 1.0  # in units of photon temperature [T_gamma]



    print("I'm calculate for:", m_a, "[eV]", T_axion, "[T_gamma]")
    
    # call CLASS
    cosmo_dict[i] = Class()
    cosmo_dict[i].set(params)
    cosmo_dict[i].set({'m_ncdm': f'{nuetrino_mass}, {m_a}'})                                            # Masses of neutino and axion
    cosmo_dict[i].set({'T_ncdm': f'{T_neutrinos}, {T_axion}'})                                          # Temperature of neutino and axion
    cosmo_dict[i].set({'ncdm_psd_parameters': f'{A(np.log(m_a))},{b(np.log(m_a))},{mu(np.log(m_a))}'})  # Axion distribution function
    cosmo_dict[i].compute()
```

```python
# finding all deltaNeff values

for i in range(0, Nums):
    derived_params = cosmo_dict[i].get_current_derived_parameters(['Neff'])  # derived parameters
    deltaNeff = derived_params['Neff'] - Neff_base
    deltaNeff_list.append(deltaNeff)

    # if i == 0:
    #     derived = cosmo_dict[i].get_current_derived_parameters(['delta_Neff'])
    #     print("Keys of derived parameters:", derived.keys())
```

```python
print("lowest value of \Delta N_eff:", min(deltaNeff_list))
```

## Show results

```python
# Find decay constants: f_a
decay_constants = calculate_fa(axion_masses)
```

```python
# Create the figure with improved aesthetics
fig, axs = plt.subplots(1, 2, figsize=(12, 6))

# ---------------------------------------------------- FIRST PART OF PLOT ---------------------------------------------------- #
# Plot the original data points
axs[0].plot(axion_masses, deltaNeff_list, color='r', label=r'$\Delta N_{\rm{eff}}$: derived from CLASS', lw=3)

# --- Other settings
# Set x-axis to log scale
axs[0].set_xscale('log')

# Labels and title with bold text
axs[0].set_xlabel('Axion Mass [eV]', fontsize=14, fontweight='bold')
axs[0].set_ylabel(r'$\bf \Delta N_{eff}$', fontsize=14, fontweight='bold')
axs[0].set_title(r'$\bf Fitting \ \Delta N_{eff} \ vs \ Axion \ Mass$', fontsize=16, fontweight='bold')
axs[0].grid(True, linestyle='--', linewidth=0.5)
axs[0].legend(fontsize=12)

# ---------------------------------------------------- SECOND PART OF PLOT ---------------------------------------------------- #
# Second plot: f_a vs Delta_N_eff with larger font size and thicker lines
axs[1].plot(decay_constants, deltaNeff_list, color='g', lw=3, label=r'$\Delta N_{\rm{eff}}$: derived from CLASS')
axs[1].set_xscale('log')
axs[1].set_xlabel(r'Decay Constant $\bf f_a$ [GeV]', fontsize=14, fontweight='bold')
axs[1].set_ylabel(r'$\bf \Delta N_{eff}$', fontsize=14, fontweight='bold')
axs[1].set_title(r'$\bf \Delta N_{eff}$ vs Decay Constant $\bf f_a$', fontsize=16, fontweight='bold')
axs[1].grid(True, linestyle='--', linewidth=0.5)
axs[1].legend(fontsize=12)

# ---------------------------------------------------- SAVING PLOT ---------------------------------------------------- #
# Adjust layout and show the plot
plt.tight_layout()
plt.show()
# plt.savefig('Fits_of_delta_N_eff_and_axion_mass.png', format='png', dpi=300)
# plt.close()
```
