```python
import numpy as np
import matplotlib.pyplot as plt
from classy import Class
```

```python
# Function to compute and plot for different sum of neutrino masses
def run_class_with_neutrino_mass_sum(sum_m_nu):
    # Divide the total neutrino mass equally among 3 neutrinos (degenerate scenario)
    m_nu_individual = sum_m_nu / 3
    
    # Define the cosmological parameters for LambdaCDM + neutrinos
    params = {
        'output': 'mPk',  # Output: matter power spectrum
        'P_k_max_1/Mpc': 3.0,  # Maximum wavenumber for the power spectrum
        'h': 0.67556,          # Reduced Hubble parameter
        'omega_b': 0.022032,   # Baryon density
        'omega_cdm': 0.12038,  # Cold dark matter density
        'n_s': 0.9619,         # Spectral index
        'A_s': 2.215e-9,       # Amplitude of scalar perturbations
        'tau_reio': 0.0925,    # Optical depth to reionization

        # Neutrino related parameters
        'N_ur': 0.046,         # Number of ultra-relativistic species (other than neutrinos)
        'N_ncdm': 1,           # Only one class of massive neutrinos
        'deg_ncdm': 3,         # 3 degenerate neutrinos
        'm_ncdm': m_nu_individual,  # Mass of each degenerate neutrino

        # Verbosity settings
        'background_verbose': 1,
        # 'input_verbose': 1,
        # 'perturbations_verbose': 2,
        # 'output_verbose': 2,
        # 'thermodynamics_verbose': 2,
    }

    # Create a CLASS instance
    cosmo = Class()

    # Set the parameters and run CLASS
    cosmo.set(params)
    cosmo.compute()

    # Retrieve the background quantities
    background = cosmo.get_background()
    # Print all keys available in the background dictionary
    print(background.keys())

    # Extract the scale factor and neutrino energy density
    z = background['z']
    R = 1 / (1 + z)
    rho_nu = background['(.)rho_ncdm[0]']  # Energy density of the neutrinos

    # Multiply by R^4 as per your requirement
    rho_nu_R4 = rho_nu * R**4

    # Plotting the result
    plt.plot(R, rho_nu_R4, label=f'$\sum m_\\nu = {sum_m_nu}$ eV')

    # Clean up CLASS instance
    cosmo.struct_cleanup()
    cosmo.empty()
```

```python
# Plot for different sum of neutrino masses
plt.figure(figsize=(8, 6))

# Run CLASS with different values for sum of neutrino masses
sum_m_nu_values = [0.0, 0.06, 0.09, 0.12]  # Different sums of neutrino masses

for sum_m_nu in sum_m_nu_values:
    run_class_with_neutrino_mass_sum(sum_m_nu)

# Finalize the plot
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Scale factor R')
plt.ylabel(r'Neutrino energy density $\rho_\nu \times R^4$ [some units]')
plt.title('Neutrino Energy Density vs Scale Factor for Different $\\sum m_\\nu$')
plt.axvline(1e-3, color='cyan', linestyle='--', label='CMB Emission')
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Divide the total neutrino mass equally among 3 neutrinos (degenerate scenario)
m_nu_individual = 0.010
sum_m_nu = m_nu_individual * 3

# Define the cosmological parameters for LambdaCDM + neutrinos
params = {
    'output': 'mPk',  # Output: matter power spectrum
    'P_k_max_1/Mpc': 3.0,  # Maximum wavenumber for the power spectrum
    'h': 0.67556,          # Reduced Hubble parameter
    'omega_b': 0.022032,   # Baryon density
    'omega_cdm': 0.12038,  # Cold dark matter density
    'n_s': 0.9619,         # Spectral index
    'A_s': 2.215e-9,       # Amplitude of scalar perturbations
    'tau_reio': 0.0925,    # Optical depth to reionization

    # Neutrino related parameters
    'N_ur': 0.0,         # Number of ultra-relativistic species (other than neutrinos)
    # 'T_ncdm': '0.71611',
    'T_ncdm': '0.68481',
    'N_ncdm': 1,           # Only one class of massive neutrinos
    'deg_ncdm': 1,         # 3 degenerate neutrinos
    'm_ncdm': m_nu_individual,  # Mass of each degenerate neutrino
    'ncdm_quadrature_strategy': '0',
    'ncdm_psd_parameters': f'{(np.log(5.0))},{(np.log(5.0))},{(np.log(5.0))}'



    # Verbosity settings
    # 'background_verbose': 5,
    # 'input_verbose': 1,
    # 'perturbations_verbose': 2,
    # 'output_verbose': 2,
    # 'thermodynamics_verbose': 2
}

# Create a CLASS instance
cosmo = Class()

# Set the parameters and run CLASS
cosmo.set(params)
cosmo.compute()
print("CLASS computation finished successfully.")  # Log if successful

# --- Retrieve the background quantities --------------------------------
background = cosmo.get_background()
# Print all keys available in the background dictionary
print(background.keys())

# --- Retrieve the derived parameters (pass the list of parameters you want) --------------------------------
derived_params = cosmo.get_current_derived_parameters(['Neff'])  # Corrected method with argument

# Extract Neff (effective number of relativistic species)
Neff = derived_params['Neff']
print(f"N_eff = {Neff}")

# --- Example: Extract the scale factor and energy density of neutrinos and axions --------------------------------
z = background['z']
R = 1 / (1 + z)
rho_nu = background['(.)rho_ncdm[0]']  # Energy density of the neutrinos

# Multiply by R^4 as per your requirement
rho_nu_R4 = rho_nu * R**4

# Plotting the result
plt.plot(R, rho_nu_R4, label=f'$\sum m_\\nu = {sum_m_nu}$ eV')

# Finalize the plot
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Scale factor R')
plt.ylabel(r'Neutrino energy density $\rho_\nu \times R^4$ [some units]')
plt.title('Neutrino Energy Density vs Scale Factor for Different $\\sum m_\\nu$')
plt.axvline(1e-3, color='cyan', linestyle='--', label='CMB Emission')
plt.legend()
plt.grid(True)
plt.show()

# Clean up CLASS instance
# cosmo.struct_cleanup()
# cosmo.empty()
```
